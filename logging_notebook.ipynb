```python
%sql
CREATE TABLE IF NOT EXISTS adb_ecommerce_prod_001.ecommerce.pipeline_logs (
    run_id STRING,
    table_name STRING,
    row_count BIGINT,
    duration_seconds INT,
    load_time TIMESTAMP
);
```

```python
# COMMAND ----------
# MAGIC %sql
# MAGIC CREATE TABLE IF NOT EXISTS ecommerce.pipeline_logs (
# MAGIC     run_id STRING,
# MAGIC     table_name STRING,
# MAGIC     row_count BIGINT,
# MAGIC     duration_seconds INT,
# MAGIC     load_time TIMESTAMP
# MAGIC );

# COMMAND ----------
dbutils.widgets.text("table_name", "")
dbutils.widgets.text("row_count", "")
dbutils.widgets.text("duration", "")
dbutils.widgets.text("run_id", "")

table_name = dbutils.widgets.get("table_name")
row_count_value = dbutils.widgets.get("row_count")
duration_value = dbutils.widgets.get("duration")
run_id = dbutils.widgets.get("run_id")

row_count = int(row_count_value) if row_count_value else 0
duration = int(duration_value) if duration_value else 0

from datetime import datetime

spark.sql(f"""
INSERT INTO ecommerce.pipeline_logs
VALUES ('{run_id}', '{table_name}', {row_count}, {duration}, '{datetime.now()}')
""")
```
